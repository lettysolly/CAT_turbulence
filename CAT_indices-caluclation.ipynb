```python
%%time
# best with one thread per worker
# takes about 12 hours with 28 cores
VARU = "wnd_ucmp"
VARV = "wnd_vcmp"
VART = "air_temp"
VARZ = "geop_ht"

from dask.utils import SerializableLock
lock = SerializableLock()


barra_vars = {"wnd_ucmp":"ua", "wnd_vcmp": "va", "air_temp":"ta", "geop_ht":"zg"}

# start and end for BARRA-R specifically
lat_min, lat_max, lon_min, lon_max = (ds_target["lat"].values[0], ds_target["lat"].values[-1], 
                                      ds_target["lon"].values[0], ds_target["lon"].values[-1])
regridder=None

def _preprocess(ds):
    ds = ds.drop_vars( ["latitude_longitude"],)\
            .sel({"latitude": slice(lat_min-0.2 , lat_max+0.2), "longitude":slice(lon_min-0.2 , lon_max+0.2)})\
            .sel({"pressure":[100, 150, 200, 250, 300,]},  method = "nearest")
            # .sel({"pressure":[p0,p,p1]},  method = "nearest")
    try:
        ds = ds.drop_vars(["forecast_period", "forecast_reference_time",])
    except:
        pass
    return ds  

with Client(threads_per_worker=1, n_workers=40) as client:
    # retry loop because sometimes this loop fails (randomly?) and can be resolved by simply retrying
    max_retries = 10
    retry_count = 0
    while retry_count < max_retries:
        try:
            ds = None
            for VAR in [VARU, VARV, VARZ, VART]:
                regridder = None
                for year in baseline_time_range:
                    new_file = f"/scratch/v46/ls7238/CAT_turbulence/BARRA-R/TMP_{barra_vars[VAR]}_AUS-15_BARRA-R_evaluation_r1i1p1f1_BOM_BARPA-R_v1-r1_6hr_{year}.nc"
                    if os.path.exists(new_file):
                        print(f"File '{new_file}' already exists.")
                        continue
                        
                    filelist = glob.glob(f"/g/data/cj37/BARRA/BARRA_R/v1/analysis/prs/{VAR}/{year}/*/{VAR}-an-prs-PT0H-BARRA_R-v1-*T*00Z.sub.nc")
                    filelist.sort()
                
                    print(f"open_mfdataset {VAR} {year}...")
                    ds = xr.open_mfdataset(filelist, preprocess=_preprocess, combine="nested", 
                            concat_dim="time", parallel=True, decode_timedelta=False, lock=lock)\
                            .rename({"latitude":"lat", "longitude":"lon", VAR:barra_vars[VAR]})\
                            .chunk({"pressure":1, "time":120, "lat":-1, "lon":-1})
    
                    if regridder is None:
                        print("calculate regridder...")
                        regridder = xe.Regridder(ds, ds_target, "bilinear",)
                    print(f"regrid and compute {year}...")
                    ds_regridded = regridder(ds)
                    # for plvl in [p0, p, p1]:
                    new_file = f"/scratch/v46/ls7238/CAT_turbulence/BARRA-R/TMP_{barra_vars[VAR]}_AUS-15_BARRA-R_evaluation_r1i1p1f1_BOM_BARPA-R_v1-r1_6hr_{year}.nc"                   
                    try:
                        print("saving netcdf...")
                        # ds_regridded.sel({"pressure":plvl}, method="nearest").to_netcdf(new_file)
                        ds_regridded.to_netcdf(new_file, engine="h5netcdf", encoding = {barra_vars[VAR]:{"zlib": True, "complevel": 6,}})
                        # ds_regridded.to_netcdf(new_file,)
                        print(f"Saved {new_file}")
                        
                        # check for any nulls
                        ds = xr.open_dataset(new_file)
                        nulls = ds.isel({"pressure":0, "lat":0, "lon":0})[barra_vars[VAR]].isnull().sum().values
                        print(f"nulls = {nulls} for {new_file}")
                        if nulls == 0:
                            pass
                        else:
                            os.remove(new_file)
                            print(f"nulls found. Removed file: {new_file}")
                            break    
                    except Exception as e:
                        # os.remove(new_file)
                        print(f"Problem saving {new_file}: {e}")
                        break
                    ds.close()
                    ds_regridded.close()  
                
        except Exception as e:
            retry_count += 1
            print(f"Error. Retry count = {retry_count}: {e}")
        break
    else:
        print(f"Operation failed after {max_retries} attempts.")
            
                    
```

# Calculate indices from standard variables

```python
def dti_preprocess(ds):
    ds = ds.drop_vars(["crs"])
    # ds = ds.dropna("time") # too slow
    ds = ds.convert_calendar("standard")
    try:
        ds = ds.expand_dims({"pressure":[int(ds.variable_id[2:])]})
    except:
        pass
    return ds
    
def delayed_turbulence_index(turbulence_index=None,
                             year=None,
                             path = "/g/data/py18/BARPA/output-CMIP6/DD/AUS-15/BOM",
                             source_id=None,
                             experiment_id=None,
                             member_id=None,
                             table_id = "6hr", 
                             version="v20231001", 
                             P=250,
                             outfile = None,
                             chunks={"time":-1, "lat":-1, "lon":-1},
                             ):
    """Use this function to compute the turbulence indices in a delayed routine"""
    # get a list of the variables needed for calculating this turbulence index (var_list)
    var_list = [f"{var}{pressure}" for var in ["ta", "ua", "va", "zg"] for pressure in [P-50, P, P+50]]
    dict_variables = {"t":"ta","u":"ua","v":"va","z":"zg",
                      "p":int(P),"p0":int(P-50),"p1":int(P+50)}
    inverted_dict = {value: key for key, value in dict_variables.items()}

    def _filename(var, pressure, source_id, experiment_id):
        VAR = f"{dict_variables[var]}{dict_variables[pressure]}"
        if source_id == "BARRA-R" and experiment_id == "evaluation":
            filename = f"/scratch/v46/ls7238/CAT_turbulence/BARRA-R/TMP_{dict_variables[var]}_AUS-15_BARRA-R_evaluation_r1i1p1f1_BOM_BARPA-R_v1-r1_6hr_{year}.nc"
        else:
            filename = f"{path}/{source_id}/{experiment_id}/{member_id}/BARPA-R/v1-r1/{table_id}/{VAR}/{version}/\
{VAR}_AUS-15_{source_id}_{experiment_id}_{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}01-{year}12.nc"
        return filename

    def _open_it(file):
        ds = xr.open_dataset(file, decode_times=True, chunks = chunks)\
            .drop_vars(["crs"], errors="ignore")\
            .astype("float32")\
            .convert_calendar("standard")

        VAR=list(ds.variables)[0]
        ds = ds.rename({VAR: inverted_dict[''.join(char for char in VAR if char.isalpha())]})
        try:
            ds = ds.expand_dims("pressure")
        except:
            pass
        ds["pressure"] = ds["pressure"].astype("int")
        return ds

    turbulence_index_vars = set()
    turbulence_index_vars.update(list(inspect.signature(globals()[turbulence_index]).parameters.keys()))
    
    params = turbulence_index_vars.intersection(["t", "u", "v", "z"])
    plvls = turbulence_index_vars.intersection(["p0", "p", "p1"])
    if len(plvls)==0:
        plvls=set(["p"])
    
    var_list = [f"{dict_variables[var]}{dict_variables[pressure]}" for var in list(params) for pressure in list(plvls)]

    
    ds = xr.merge([xr.concat([_open_it(_filename(var, pressure, source_id, experiment_id)).sel({"pressure":dict_variables[pressure]}, method="nearest") 
                              for pressure in list(plvls)], 
                             dim="pressure") 
                   for var in list(params)], join='outer')
    
    # delayed write to file. Compute outside this function
    
    # lazy calculate indices
    ds = calc_turbulence_indices(ds, which= turbulence_index, p=P, u="u", v="v", t="t", z="z",)

    if outfile is None:
        outfile = f"/scratch/v46/ls7238/CAT_turbulence/TMP_{turbulence_index}/TMP_{turbulence_index}-{P}hPa_AUS-15_{source_id}_{experiment_id}_{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}.nc"

    ds_to_write = ds[[turbulence_index]].sel({"time":str(year)})
    try:
        ds_to_write = ds_to_write.sel({"pressure":P}, method="nearest",)
    except:
        pass
    delayed_write = ds_to_write.to_netcdf(outfile, mode="a", compute=False,)
    return delayed_write
```

```python
# for all scenarios, calculate the index, then calculate the monthly frequency above the BARRA-R baseline 99th percentile
# for baseline periods, also calculate the percentiles 

# output saved to new subfolder within original folders: /scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/percentiles
# and /scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/freq-above-p99
# TMP files removed 

# best with one worker, many threads for barra
# use 28 cores 4 threads, 2 workers

new_folder = "mmm-p99"

dict_years = {"evaluation":(1990, 2009), 
              "historical":(1979, 2014), 
              "future":(2015, 2100)}
dict_model_index = {"evaluation":["evaluation_BARRA-R_r1i1p1f1"], 
                    "historical":list_historical,
                    "future":list_future}

def p99freq_preprocess(ds):
    """Calculate annual frequency of exceeding p99 threshold"""
    return (ds < p99).resample({"time":"ME"},).mean(["time"], skipna=True)    
    # change <p99 for neg var (Ri, AbsVort), >p99 for positive (all others)

def quantiles_preprocess(ds):
    """For evaluation from historical, calculate the monthly values of 1st to 99th quantiles within the mid lat box"""
    ds = ds.convert_calendar("standard")
    # rechunk such that there are as many chunks as there are years, 
    ds = ds.chunk({"time": 4*365, "lat": -1, "lon": -1})
    ds = ds.sel(lat=mid_lat_slice, lon=lon_slice)\
            .resample(time="ME")\
            .apply(lambda ds: ds.quantile(np.arange(0,1,0.01), dim=["lat", "lon", "time"]))
    return ds
    
def run_p99freq_and_quantiles(scenario, chunks={"time":-1, "lat":-1, "lon":-1}):
    """expects scenario as one of  [ "historical", "future", "evaluation", ]"""
    for run in dict_model_index[scenario]:
        experiment_id, source_id, member_id = run.split("_")   

        os.makedirs(f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}", exist_ok=True)
        os.makedirs(f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/freq-above-p99/", exist_ok=True)
        p99_filename = f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/freq-above-p99/{turbulence_index}-{P}hPa-monthly-freq-above-p99_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"
        
        if scenario =="historical" or scenario=="evaluation":
            frequency = "monthly"
            perc_filename = f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/percentiles/{turbulence_index}-{P}hPa-{frequency}-percentiles_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"
            perc_filename_exists = os.path.exists(perc_filename)
        else:
            perc_filename_exists = True
        
        if os.path.exists(p99_filename) and perc_filename_exists:
            print(f"{p99_filename} already exists, skipping {run}")
            continue

        delayed_list=[]
        start_year, end_year = dict_years[scenario]
        for year in np.arange(start_year, end_year+1):
            tmp_file = f"/scratch/v46/ls7238/CAT_turbulence/TMP_{turbulence_index}/TMP_{turbulence_index}-{P}hPa_AUS-15_{source_id}_{experiment_id}_{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}.nc"
            if os.path.exists(tmp_file):
                print(f"File '{tmp_file}' already exists.")
            else:
                print(f"Making {tmp_file}")
                delayed_list.append(delayed_turbulence_index(turbulence_index=turbulence_index,
                                                      year=year,
                                                      source_id=source_id,
                                                      experiment_id=experiment_id,
                                                      member_id=member_id,
                                                             P=P,
                                                             outfile=tmp_file,
                                                             chunks=chunks,
                                                            ))
                
        if len(delayed_list)>0:
            print("compute ... ")
            #dask.compute(*delayed_list)
            batch_size = 5
            for i in range(0, len(delayed_list), batch_size):
                batch = delayed_list[i:i+batch_size]
                print(f"Computing batch {i//batch_size + 1} of {(len(delayed_list)-1)//batch_size + 1}...")
                dask.compute(*batch)

        # compute freq
        # multi years
        filelist = [f"/scratch/v46/ls7238/CAT_turbulence/TMP_{turbulence_index}/TMP_{turbulence_index}-{P}hPa_AUS-15_{source_id}_{experiment_id}_{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}.nc" 
                    for year in np.arange(start_year, end_year+1)]
        
        if os.path.exists(p99_filename):
            print(f"File '{p99_filename}' already exists.")
            
        else:
            # calculate monthly freq above p99
            ds =  xr.open_mfdataset(filelist,
                                   preprocess=p99freq_preprocess, 
                                   combine="nested", 
                                   concat_dim="time",
                                   )\
                .assign_coords({"run":run})\
                .assign_attrs({"turbulence_index": turbulence_index,
                               "pressure_level": P,
                               "Description": f"Frequency of {turbulence_index} at {P}hPa above the 99th percentile for BARPA-R experiments based on BARPA-R MMM percentile from 1990 to 2009 within latitudes (-50 to -25) and longitudes (90 to 195) using 6-hourly data and calculating frequencies per calendar month",
                               
                               "p99":p99,})\
                .to_netcdf(p99_filename, compute=True)
            print(f"made {p99_filename}")

        if scenario =="historical" or scenario=="evaluation":
            # calculate quantiles for evaluation period only
            frequency = "monthly"
            os.makedirs(f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/percentiles", exist_ok=True)
            perc_filename = f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/percentiles/{turbulence_index}-{P}hPa-{frequency}-percentiles_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"
            if os.path.exists(perc_filename):
                print(f"File '{perc_filename}' already exists.")
            else:
                filelist_baseline = [f"/scratch/v46/ls7238/CAT_turbulence/TMP_{turbulence_index}/TMP_{turbulence_index}-{P}hPa_AUS-15_{source_id}_{experiment_id}_{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}.nc"
                            for year in baseline_time_range]
                
                ds = xr.open_mfdataset(filelist_baseline, 
                                       combine="nested", 
                                       concat_dim="time",
                                       preprocess=quantiles_preprocess,
                                      )\
                        .assign_coords({"run":run})\
                        .assign_attrs({"turbulence_index": turbulence_index,
                                       "pressure_level": P,
                                       "Description": f"Percentiles of {turbulence_index} at {P}hPa for BARPA-R experiments using 6-hourly data and calculating frequencies per calendar month for years 1990 to 2009",
                                      })
                null= ds.isnull()[turbulence_index].sum().values
                # assert null==0, print(f"Number of null values: {null} in {run}.")
                
                try:
                    ds.to_netcdf(perc_filename, compute=True)
                    print(f"Made '{perc_filename}'")
                except Exception as e:
                    print(f"Error in {run}: {e}")


        
        
        # Saving 10 case studies where 6 hour frequency is exceeded the most across the 
        # domain and where daily mean exceedence is the greatest across the domain
        case_study_idx = ["VWS", "TI1", "windspeed"]
        if scenario == "historical" and turbulence_index in case_study_idx:
            case_study_dir = f"/scratch/v46/ls7238/CAT_turbulence/{turbulence_index}/{P}hPa/{new_folder}/case-studies"
            os.makedirs(case_study_dir, exist_ok=True)

            # Keep file naming convention similar to Gen's
            top10_6hr_filename = f"{case_study_dir}/{turbulence_index}-{P}hPa-top10-6hr-days_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"
            top10_daily_filename = f"{case_study_dir}/{turbulence_index}-{P}hPa-top10-daily-mean_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"

            if os.path.exists(top10_6hr_filename) and os.path.exists(top10_daily_filename):
                print(f"Both already exist: {top10_6hr_filename} and {top10_daily_filename}, skipping")
            
            else:
                # open full historical TMP files
                ds_hist = xr.open_mfdataset(filelist, combine="nested", concat_dim="time")
                da = ds_hist[turbulence_index]
                exceed = (da > p99).sel(time=slice("1990", None))

                # Output 1: top 10 days by peak 6-hourly exceedance count
                if not os.path.exists(top10_6hr_filename):
                    # count grid points exceeding p99 at each 6-hourly timestep
                    count_6hr = exceed.sum(dim=["lat", "lon"]).compute()
                    top10_times = pd.Series(count_6hr.values, 
                                            index=count_6hr.time.values).nlargest(10).index

                    # save the raw TI1 field for the full day containing each top timestep
                    day_list = []
                    for rank, t in enumerate(top10_times):
                        date_str = pd.Timestamp(t).strftime('%Y-%m-%d')
                        day_data = da.sel(time=date_str).compute()
                        day_data = day_data.expand_dims({"event": [rank + 1]})
                        day_data = day_data.assign_coords({
                            "peak_timestep": ("event", [t]),
                            "peak_count":    ("event", [int(count_6hr.sel(time=t).values)])
                        })
                        day_list.append(day_data)

                    xr.concat(day_list, dim="event")\
                    .to_dataset(name=turbulence_index)\
                    .assign_attrs({"Description": f"Top 10 days (full 6-hourly) by peak 6-hourly exceedance count of {turbulence_index} > {p99:.4e} for {run}",
                                    "p99": p99})\
                    .to_netcdf(top10_6hr_filename)
                    print(f"Saved {top10_6hr_filename}")

                # Output 2: top 10 days by daily mean exceedance count
                if not os.path.exists(top10_daily_filename):
                    # daily mean exceedance field (fraction of 6-hourly timesteps exceeding p99)
                    daily_exceed = exceed.resample(time="1D").mean()
                    # sum over domain to rank days
                    daily_count = daily_exceed.sum(dim=["lat", "lon"]).compute()
                    top10_day_series = pd.Series(daily_count.values, 
                                                index=daily_count.time.values).nlargest(10)
                    top10_days = top10_day_series.index

                    top10_daily_fields = daily_exceed.sel(time=top10_days).compute()

                    top10_daily_fields\
                    .to_dataset(name=turbulence_index)\
                    .assign_coords({
                        "rank":        ("time", np.arange(1, 11)),
                        "daily_count": ("time", top10_day_series.values)
                    })\
                    .assign_attrs({"Description": f"Top 10 days by daily mean exceedance of {turbulence_index} > {p99:.4e} for {run}",
                                    "p99": p99})\
                    .to_netcdf(top10_daily_filename)
                    print(f"Saved {top10_daily_filename}")
        


        # check that freq calculated correctly
        # then delete all in filelist
        for file in filelist:
            if os.path.exists(file):
                os.remove(file)
                print(f"File removed: {file}")
            else:
                print(f"File does not exist: {file}")
                pass
    return
```

```python
# Windspeed case study calculation (full calc was done before case study addition)

import os

# Pin BLAS threading to 1 so dask's thread count is the only parallelism level
os.environ["OMP_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["NUMEXPR_NUM_THREADS"] = "1"

import sys
import inspect
import warnings
import numpy as np
import pandas as pd
import xarray as xr
import dask
from dask.distributed import Client

from cat_indices import calc_turbulence_indices, windspeed, VWS, TI1, AbsVort, Ri, TI2, TI3

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

import logging
logging.getLogger("flox").setLevel(logging.WARNING)

# --- EDIT THESE to match your ARE session's real allocation ---
ncpus = 28
memory_limit = "115GB"   # keep comfortably below your session's actual cgroup limit
# ----------------------------------------------------------------

turbulence_index = "windspeed"
run_tag = turbulence_index  # no _TEST suffix — this is a real run

P = 250
new_folder = "mmm-p99"
start_year, end_year = 1990, 2014
BATCH_SIZE = 10  # windspeed is light enough this can likely go higher than 5

list_historical = [
    "historical_ACCESS-CM2_r4i1p1f1",
    "historical_ACCESS-ESM1-5_r6i1p1f1",
    "historical_CESM2_r11i1p1f1",
    "historical_CMCC-ESM2_r1i1p1f1",
    "historical_EC-Earth3_r1i1p1f1",
    "historical_MPI-ESM1-2-HR_r1i1p1f1",
    "historical_NorESM2-MM_r1i1p1f1",
]

mmm_p99_dict = {
    "windspeed": 76.7527, "VWS": 0.012295220450099018, "Ri": 1.152088,
    "TI1": 2.1185820196367124e-07, "TI2": 2.267155e-07, "TI3": 2.3883844807661087e-07,
}
p99 = mmm_p99_dict[turbulence_index]

def delayed_turbulence_index(
    turbulence_index=None, year=None,
    path="/g/data/py18/BARPA/output-CMIP6/DD/AUS-15/BOM",
    source_id=None, experiment_id=None, member_id=None,
    table_id="6hr", version="v20231001", P=250, outfile=None,
    chunks={"time": -1, "lat": -1, "lon": -1},
):
    dict_variables = {
        "t": "ta", "u": "ua", "v": "va", "z": "zg",
        "p": int(P), "p0": int(P - 50), "p1": int(P + 50),
    }
    inverted_dict = {value: key for key, value in dict_variables.items()}

    def _filename(var, pressure, source_id, experiment_id):
        VAR = f"{dict_variables[var]}{dict_variables[pressure]}"
        if source_id == "BARRA-R" and experiment_id == "evaluation":
            return (f"/scratch/v46/ls7238/CAT_turbulence/BARRA-R/"
                    f"TMP_{dict_variables[var]}_AUS-15_BARRA-R_evaluation_r1i1p1f1_"
                    f"BOM_BARPA-R_v1-r1_6hr_{year}.nc")
        return (f"{path}/{source_id}/{experiment_id}/{member_id}/BARPA-R/v1-r1/"
                f"{table_id}/{VAR}/{version}/"
                f"{VAR}_AUS-15_{source_id}_{experiment_id}_{member_id}_"
                f"BOM_BARPA-R_v1-r1_6hr_{year}01-{year}12.nc")

    def _open_it(file):
        ds = (xr.open_dataset(file, decode_times=True, chunks=chunks)
              .drop_vars(["crs"], errors="ignore").astype("float32")
              .convert_calendar("standard"))
        VAR = list(ds.variables)[0]
        ds = ds.rename({VAR: inverted_dict["".join(c for c in VAR if c.isalpha())]})
        try:
            ds = ds.expand_dims("pressure")
        except Exception:
            pass
        ds["pressure"] = ds["pressure"].astype("int")
        return ds

    turbulence_index_vars = set(inspect.signature(globals()[turbulence_index]).parameters.keys())
    params = turbulence_index_vars.intersection(["t", "u", "v", "z"])
    plvls = turbulence_index_vars.intersection(["p0", "p", "p1"])
    if len(plvls) == 0:
        plvls = set(["p"])

    ds = xr.merge(
        [xr.concat([_open_it(_filename(var, pressure, source_id, experiment_id))
                    .sel({"pressure": dict_variables[pressure]}, method="nearest")
                    for pressure in list(plvls)], dim="pressure")
         for var in list(params)], join="outer",
    )
    ds = calc_turbulence_indices(ds, which=turbulence_index, p=P, u="u", v="v", t="t", z="z")

    if outfile is None:
        outfile = (f"/scratch/v46/ls7238/CAT_turbulence/TMP_{turbulence_index}/"
                   f"TMP_{turbulence_index}-{P}hPa_AUS-15_{source_id}_{experiment_id}_"
                   f"{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}.nc")

    ds_to_write = ds[[turbulence_index]].sel({"time": str(year)})
    try:
        ds_to_write = ds_to_write.sel({"pressure": P}, method="nearest")
    except Exception:
        pass
    return ds_to_write.to_netcdf(outfile, mode="a", compute=False)

def run_case_study_for(run):
    experiment_id, source_id, member_id = run.split("_")

    case_study_dir = f"/scratch/v46/ls7238/CAT_turbulence/{run_tag}/{P}hPa/{new_folder}/case-studies"
    os.makedirs(case_study_dir, exist_ok=True)
    top10_6hr_filename = f"{case_study_dir}/{turbulence_index}-{P}hPa-top10-6hr-days_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"
    top10_daily_filename = f"{case_study_dir}/{turbulence_index}-{P}hPa-top10-daily-mean_AUS-15_{run}_BOM_BARPA-R_v1-r1_6hr.nc"

    if os.path.exists(top10_6hr_filename) and os.path.exists(top10_daily_filename):
        print(f"Both already exist for {run}, skipping")
        return

    tmp_dir = f"/scratch/v46/ls7238/CAT_turbulence/TMP_{run_tag}"
    os.makedirs(tmp_dir, exist_ok=True)

    filelist, delayed_list = [], []
    for year in np.arange(start_year, end_year + 1):
        tmp_file = (f"{tmp_dir}/TMP_{run_tag}-{P}hPa_AUS-15_{source_id}_{experiment_id}_"
                    f"{member_id}_BOM_BARPA-R_v1-r1_6hr_{year}.nc")
        filelist.append(tmp_file)
        if os.path.exists(tmp_file):
            print(f"File '{tmp_file}' already exists.")
        else:
            print(f"Making {tmp_file}")
            delayed_list.append(delayed_turbulence_index(
                turbulence_index=turbulence_index, year=year, source_id=source_id,
                experiment_id=experiment_id, member_id=member_id, P=P, outfile=tmp_file,
            ))

    if delayed_list:
        print("compute ... ")
        for i in range(0, len(delayed_list), BATCH_SIZE):
            batch = delayed_list[i:i + BATCH_SIZE]
            print(f"Computing batch {i // BATCH_SIZE + 1} of {(len(delayed_list) - 1) // BATCH_SIZE + 1}...")
            dask.compute(*batch)

    ds_hist = xr.open_mfdataset(filelist, combine="nested", concat_dim="time", parallel=True)
    da = ds_hist[turbulence_index]
    exceed = (da > p99).sel(time=slice("1990", None))

    if not os.path.exists(top10_6hr_filename):
        count_6hr = exceed.sum(dim=["lat", "lon"]).compute()
        top10_times = pd.Series(count_6hr.values, index=count_6hr.time.values).nlargest(10).index
        day_list = []
        for rank, t in enumerate(top10_times):
            day_data = da.sel(time=pd.Timestamp(t).strftime("%Y-%m-%d")).compute()
            day_data = day_data.expand_dims({"event": [rank + 1]}).assign_coords({
                "peak_timestep": ("event", [t]),
                "peak_count": ("event", [int(count_6hr.sel(time=t).values)]),
            })
            day_list.append(day_data)
        xr.concat(day_list, dim="event").to_dataset(name=turbulence_index).assign_attrs({
            "Description": f"Top 10 days (full 6-hourly) by peak 6-hourly exceedance count of {turbulence_index} > {p99:.4e} for {run}",
            "p99": p99,
        }).to_netcdf(top10_6hr_filename)
        print(f"Saved {top10_6hr_filename}")

    if not os.path.exists(top10_daily_filename):
        daily_exceed = exceed.resample(time="1D").mean()
        daily_count = daily_exceed.sum(dim=["lat", "lon"]).compute()
        top10_day_series = pd.Series(daily_count.values, index=daily_count.time.values).nlargest(10)
        top10_daily_fields = daily_exceed.sel(time=top10_day_series.index).compute()
        top10_daily_fields.to_dataset(name=turbulence_index).assign_coords({
            "rank": ("time", np.arange(1, 11)),
            "daily_count": ("time", top10_day_series.values),
        }).assign_attrs({
            "Description": f"Top 10 days by daily mean exceedance of {turbulence_index} > {p99:.4e} for {run}",
            "p99": p99,
        }).to_netcdf(top10_daily_filename)
        print(f"Saved {top10_daily_filename}")

    for file in filelist:
        if os.path.exists(file):
            os.remove(file)
            print(f"File removed: {file}")
```
